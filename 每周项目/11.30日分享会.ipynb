## 2. JSON 数据格式处理

### 2.1 什么是 JSON？
- 轻量级的数据交换格式
- 常用于前后端数据传输、配置文件等

```python
import json

# Python字典
data = {
    "姓名": "小红",
    "年龄": 18,
    "成绩": [85, 92, 78]
}
print(data)
# 转为JSON字符串
json_str = json.dumps(data, ensure_ascii=False)
print("JSON字符串:")
print(json_str)

# 保存到文件
with open("data.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

# 从文件读取
with open("data.json", "r", encoding="utf-8") as f:
    loaded_data = json.load(f)

print("\n读取的数据:")
print(f"姓名: {loaded_data['姓名']}")
print(f"年龄: {loaded_data['成绩']}")
```

### 2.2 使用 `json` 模块  
- `json.loads()`：将 JSON 字符串转为 Python 对象  
- `json.dumps()`：将 Python 对象转为 JSON 字符串  
- `json.load()`：从文件读取 JSON  
- `json.dump()`：将 JSON 写入文件

```python
import json

# 字典转 JSON 字符串
data = {"name": "小明", "age": 18, "hobby": ["篮球", "音乐"]}
json_str = json.dumps(data, ensure_ascii=False)
print(json_str)

# 写入 JSON 文件
with open("data.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=4)

# 读取 JSON 文件
with open("data.json", "r", encoding="utf-8") as f:
    loaded_data = json.load(f)
    print(loaded_data)
```

## 3. 异常捕获与处理

### 3.1 异常类型举例
- `ZeroDivisionError`：除零错误  
- `FileNotFoundError`：文件不存在  
- `ValueError`：值错误  
- `KeyError`：字典键不存在  
- `Exception`：所有异常的基类

### 3.2 使用 `try...except...finally`

```python
a=1
try:
    num = int(input("请输入一个数字："))
    result = 10 / num
    print(f"结果是：{result}")
except ZeroDivisionError:
    print("不能除以零！")
except ValueError:
    print("请输入有效数字！")
finally:
    print("程序执行完毕")
```
