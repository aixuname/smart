## 2.变量数据类型:
### 整型(int):整数,例如:1,2,3,4,5......

## 条件语句(if - elif -else)
### 基本结构

```python
if "条件":
    "要执行的代码"
elif "条件":        
     "要执行的代码"
else:        
     "要执行的代码"  
```

```python
#例子
import random
a=random.randint(1,1000)
if a%2==0:
    print(f"{a}为偶数")
else :
    print(f"{a}为奇数")
```

```python
#条件表达式
#a>b
#a==b
#a and b
#a>b and c>b 
#条件语句本身是判断条件的结果是否为True 或 1
#若是True或1则会执行后续的代码,否则就进行下一个条件的判断
#当然也可以直接输入True ,如果要执行的代码只有1行则可以不用换行,直接在引号的后面写出即可
if True:print("我爱学习")
else :print("我不爱学习")
```

### 嵌套条件语句

```python
if "条件":
    if "条件":
        "执行的代码"
    else:
        "执行的代码"
else:
    "执行的代码"
```

## 循环语句(for,while)
### for循环基本结构

```python
for a in range(666):
    print("哈基米哦南北绿豆")
```

### 上述的a是一个变量
### range()为一个序列,range(始,末,步长)
### 当"始"的值如果不输入默认为0,步长默认为1
### range的取值区间为左闭右开,即对于range(1,6)序列来说取值为1,2,3,4,5
### 当括号里只有2个数字时则对应是"始"和"末",如range(1,5),只有一个数是则是"末",如range(5)

```python
#起始为1,结束为5,步长为2
for i in range(1,5):
    print("哈基米哦南北绿豆")
```

### for 循环还可以遍历字符串

```python
for i in "明月几时有":
    print(i)
```

### 条件循环while基本结构

```python
a=0
while True:
    a+=1
    print(f"{a}哈基米哦南北绿豆")
```

## continue 和 break
### continue是跳过这次循环,break是立即退出循环

```python
#只打印奇数
count = 0
while count < 10:
    count += 1
    if count % 2 == 0:  
        continue 
    print(f"奇数: {count}")
```
